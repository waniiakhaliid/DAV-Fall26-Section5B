```python
import pandas as pd

df = pd.read_csv("adult.csv")

print(df.head())
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.barplot(data=df, x="education", y="hours.per.week", palette="pastel")

plt.xlabel("Education Level")
plt.ylabel("Average Hours per Week")
plt.title("Average Working Hours by Education Level")
plt.xticks(rotation=90)
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.barplot(data=df, x="marital.status", y="hours.per.week", color="lightblue")

plt.xlabel("Marital Status")
plt.ylabel("Average Hours per Week")
plt.title("Average Working Hours by Marital Status")
plt.xticks(rotation=45)
plt.show()
```

```python
# Task 2 — Frequency of each education category
sns.countplot(data=df, x="education", palette="pastel")

plt.xlabel("Education Level")
plt.ylabel("Number of Records")
plt.title("Number of People by Education Level")
plt.xticks(rotation=90)
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.histplot(data=df, x="age", bins=20, color="pink")

plt.xlabel("Age")
plt.ylabel("Frequency")
plt.title("Distribution of Age")
plt.show()
```

```python
# task 4 — Box plot of age
import seaborn as sns
import matplotlib.pyplot as plt

sns.boxplot(data=df, y="age", color="pink")

plt.ylabel("Age")
plt.title("Box Plot of Age")
plt.show()
```

```python
# task 5
Q1 = df["age"].quantile(0.25)
Q3 = df["age"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[(df["age"] < lower) | (df["age"] > upper)]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower)
print("Upper Bound:", upper)
print("Number of Outliers:", len(outliers))
```

```python
# task 6
from scipy import stats

df["zscore"] = stats.zscore(df["age"])

outliers_z = df[df["zscore"].abs() > 3]

print("Number of Z-score Outliers:", len(outliers_z))
```

```python
# task 7
import seaborn as sns
import matplotlib.pyplot as plt

sns.scatterplot(
    data=df,
    x="age",
    y="hours.per.week",
    hue="income"
)

plt.xlabel("Age")
plt.ylabel("Hours per Week")
plt.title("Age vs Hours Worked per Week")
plt.show()
```

```python
# task 8
avg_age = df.groupby("education.num")["age"].mean()

plt.plot(avg_age.index, avg_age.values, marker="o")

plt.xlabel("Education Number")
plt.ylabel("Average Age")
plt.title("Average Age Across Education Levels")
plt.show()
```

```python
plt.fill_between(avg_age.index, avg_age.values, alpha=0.3)
plt.plot(avg_age.index, avg_age.values)

plt.xlabel("Education Number")
plt.ylabel("Average Age")
plt.title("Average Age Across Education Levels")
plt.show()
```

```python
# task 9
import seaborn as sns
import matplotlib.pyplot as plt

avg_hours = df.groupby("marital.status")["hours.per.week"].mean()

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)

sns.barplot(
    x=avg_hours.index,
    y=avg_hours.values,
    color="pink"
)

plt.xlabel("Marital Status")
plt.ylabel("Average Hours per Week")
plt.title("Normal Bar Chart")
plt.xticks(rotation=45)
```

```python
plt.subplot(1, 2, 2)

sns.barplot(
    x=avg_hours.index,
    y=avg_hours.values,
    color="pink"
)

plt.ylim(bottom=35)

plt.xlabel("Marital Status")
plt.ylabel("Average Hours per Week")
plt.title("Truncated Y-Axis")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```
